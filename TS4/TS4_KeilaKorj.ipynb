## ***Introducción***

Partiendo de una transferencia que tiene por objetivo simplemente aportar un retardo de fase $t_o$:
$$
T(j \omega)= |T(\omega)| \cdot e^{-j \omega t_o} = |T| \cdot e^{- \frac{j \omega}{\frac{1}{t_o}}}   
$$

Donde $\Omega_w = \frac{1}{t_o}$, de modo que:
$$
T(j \omega')= |T(\omega')| \cdot e^{-j \omega'}
$$

Es así como reemplazando $\omega' = \frac{s}{j}$ se obtiene:
$$
T(s) = |T|\cdot e^{-j\cdot \frac{s}{j}} = |T| \cdot e^{-s} = |T|\cdot e^{\frac{1}{s}}
$$

Por otro lado, siendo:
$$
i) \space e^{s} + e^{-s} = 2\cdot \operatorname{cosh}(s)
ii) \space e^{s} - e^{-s} = 2\cdot \operatorname{senh}(s)
$$

Si se hace (i + ii):
$$
e^{s} = \operatorname{cosh}(s) + \operatorname{senh}(s)
$$

Con lo cual, volviendo a la transferencia y tomando a $|T| = 1$:
$$
T(s)=|T|\cdot e^{\frac{1}{s}} = \frac{1}{\operatorname{cosh}(s) + \operatorname{senh}(s)}
$$

Si se descompone a $\operatorname{cosh}(s)$ y $\operatorname{senh}(s)$ por serie de Taylor se obtiene:
$$
i) \operatorname{senh}(s) = s + \frac{s^{3}}{3!} + \frac{s^{5}}{5!}+...
$$
$$
ii) \operatorname{cosh}(s) = s + \frac{s^{2}}{2!} + \frac{s^{4}}{4!}+...
$$

Y tomando en cuenta a:
$$
\operatorname{cosh}(s) = \frac{\operatorname{cosh}(s)}{\operatorname{senh}(s)}
$$
Se permite de este modo una expansión sucesiva continua que lleva a:
$$
\operatorname{cotgh}(s) = \frac{\operatorname{cosh}(s)}{\operatorname{senh}(s)} = \frac{1}{s} \cdot \frac{1}{\frac{3}{s} + \frac{1}{\frac{5}{s}+ \frac{1}{\frac{7}{s} + ...}}}
$$

Por lo tanto de esta manera obtenemos las expresiones de $\operatorname{senh}(s)$ y $\operatorname{cosh}(s)$ y así adquirimos el polinomio del denominador de la transferencia de Bessel para los distintos ordenes $n$. Tal que de modo general la expresión es:
$$
T_{B_n}(s) = \frac{B_{n_0}}{B_n(s)}
$$
Siendo $B_{n_{0}}$ el térmmino independiente del polinomio del denominador.


## ***Transferencia T(s) de Bessel para $n=2,3$ y $4$ normalizados para $D(\omega) =1s$***

Tomando como base la introducción provista, se procede de la siguiente forma:

### *Orden 2*

Al buscar la transferencia de orden $n=2$, entonces:

$$
\operatorname{cotgh}(s) = \frac{1}{s} \cdot \frac{1}{\frac{3}{s}} = \frac{1}{s} \cdot \frac{1}{\frac{s}{3}}
$$
Quedando:
$$
\operatorname{cotgh}(s) = \frac{s^{2}+3}{3\cdot s} = \frac{\operatorname{cosh}(s)}{\operatorname{senh}(s)}
$$
Por lo tanto:
$$
\operatorname{cosh}(s) = s^{2}+3
\operatorname{senh}(s) = 3\cdot s
$$

Resultando en:
$$
B_n(s)=\operatorname{cosh}(s) + \operatorname{senh}(s)
$$
$$
B_{n_{0}} = 3
$$

Para finalmente:
$$
T_{B_{2}}(s) = \frac{3}{s^{2} + 3\cdot s + 3} 
$$

### *Orden 3*

Al buscar la transferencia de orden $n=3$, entonces:

$$
\operatorname{cotgh}(s) = \frac{1}{s} \cdot \frac{1}{\frac{3}{s} + \frac{1}{\frac{5}{s}}} = \frac{1}{s} \cdot \frac{1}{\frac{s}{3} + \frac{s}{5}}
$$
Quedando:
$$
\operatorname{cotgh}(s) = \frac{6\cdot s^{2} + 15}{s^{3} + 15\cdot s} = \frac{\operatorname{cosh}(s)}{\operatorname{senh}(s)}
$$
Por lo tanto:
$$
\operatorname{cosh}(s) = 6\cdot s^{2} + 15
\operatorname{senh}(s) = s^{3} + 15\cdot s
$$

Resultando en:
$$
B_n(s)=\operatorname{cosh}(s) + \operatorname{senh}(s)
$$
$$
B_{n_{0}} = 15
$$

Para finalmente:
$$
T_{B_{3}}(s) = \frac{15}{s^{3} + 6 \cdot s^{2} + 15\cdot s + 15} 
$$

### *Orden 4*

Al buscar la transferencia de orden $n=4$, entonces:

$$
\operatorname{cotgh}(s) = \frac{\operatorname{cosh}(s)}{\operatorname{senh}(s)} = \frac{1}{s} \cdot \frac{1}{\frac{3}{s} + \frac{1}{\frac{5}{s}+ \frac{1}{\frac{7}{s}}}}
$$

Quedando:
$$
\operatorname{cotgh}(s) = \frac{s^{4} + 45 \cdot s^{2} + 105}{10 \cdot s^{3} + 105 \cdot s} = \frac{\operatorname{cosh}(s)}{\operatorname{senh}(s)}
$$

Por lo tanto:
$$
\operatorname{cosh}(s) = s^{4} + 45 \cdot s^{2} + 105
\operatorname{senh}(s) = 10 \cdot s^{3} + 105 \cdot s
$$

Resultando en:
$$
B_n(s)=\operatorname{cosh}(s) + \operatorname{senh}(s)
B_{n_{0}} = 105
$$

Para finalmente:
$$
T_{B_{4}}(s) = \frac{105}{s^{4} + 10 \cdot s^{3} + 45 \cdot s^{2} + 105 \cdot s} + 105 
$$

## ***Evaluar el retardo de grupo $D(\omega = 2.5)$ y expresar en forma porcentual [%] el error o desviamiento respecto a $D(\omega = 0)$***

Desde:
$$
T_{B_{3}}(s) = \frac{15}{s^{3} + 6 \cdot s^{2} + 15\cdot s + 15} 
$$

Se busca $T(jw)$, por ende $s=jw$, con lo cual:
$$
T(j\omega) = \frac{15}{-j\cdot \omega^{3} - 6 \cdot \omega^{2} + j\cdot15\cdot \omega  + 15}  = \frac{15}{(- 6 \cdot \omega^{2} +15) + j\cdot(15\cdot \omega - \omega^{3})}
$$

Si se divide la transferencia en módulo y fase $T(j\omega) = |T| \cdot e^{j\angle T}$:

$$
T(jw) = \frac{15}{\sqrt{(-6\omega^{2} + 15)^{2} + (- \omega^{3} + 15\omega)^{2}}} \cdot e^{-j \operatorname{tg}^{-1}(\frac{- \omega^{3} + 15\omega}{-6\omega^{2} + 15})}
$$

Con lo cual:
$$
\angle T = -\operatorname{tg}^{-1}(\frac{- \omega^{3} + 15\omega}{-6\omega^{2} + 15})
$$
Y siendo:
$$
D(\omega) = - \frac{d\angle T}{d\omega} = \frac{d(\operatorname{tg}^{-1}(\frac{- \omega^{3} + 15\omega}{-6\omega^{2} + 15}))}{d\omega}
$$

Quedando finalmente:
$$
D(\omega) = \frac{6\cdot w^{4}  + 45 \cdot s^{2} + 225}{w^{6} + 6\cdot w^{4}  + 45 \cdot s^{2} + 225}
$$

De modo que:
$$
D(\omega = 0) = 1s
$$
y

$$
D(\omega = 2.5) = 0.7520s
$$

Siendo este útlimo resultado el $75.2\%$ del valor de $D(\omega = 0)$, se puede decir que el error porcentual es:
$$
\varepsilon_{\%} = 100\% - \frac{D(\omega = 2.5) . 100}{D(\omega = 0)} = 24.8\%
$$

## ***Implementantación circuital normalizada utilizando estructuras Sallen-Key con K=1*** 

Para ello se necesita obtener la expresión de la transferencia dividida en secciones, y para ello se utiliza:

```python
# Inicialización e importación de módulos

# Módulos externos
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import scipy.signal as sig

fig_sz_x = 13
fig_sz_y = 7
fig_dpi = 80 # dpi

fig_font_size = 11

mpl.rcParams['figure.figsize'] = (fig_sz_x, fig_sz_y)
mpl.rcParams['figure.dpi'] = fig_dpi
plt.rcParams.update({'font.size':fig_font_size})

from pytc2.sistemas_lineales import analyze_sys, pretty_print_bicuad_omegayq, tf2sos_analog, pretty_print_SOS, pretty_print_lti

from pytc2.general import print_subtitle


this_order = 3

z,p,k = sig.besselap(this_order, norm='delay')

num, den = sig.zpk2tf(z,p,k)

lp_bessel = tf2sos_analog(num, den)

print("LP Bessel Orden 3")

pretty_print_lti(num, den)

print("LP Bessel Orden 3 dividida en secciones")

pretty_print_SOS(lp_bessel)
```

Donde puede observarse uns sección de primer orden y otra de segundo orden, respetando el orden $n=3$ del filtro.

### *Sección de 1° orden*

Aquí se hizo uso de un circuito RC serie tal que la transferencia es:
$$
T_{RC}(s) = \frac{RC}{s + \frac{1}{RC}}
$$
Tomando como norma de impedancia $\Omega_z = R$, queda:
$$
T(s) = \frac{C'}{s + \frac{1}{C'}} = \frac{2.322}{s + 2.322}
$$
Por ende:
$$
\frac{1}{C'} = 2.322
$$
Para finalmente obtener:

$$
C' = 0.4306
$$
$$
R'=1
$$

### ***Desnormalización de los componentes para obtener un $D(w=0) = 200 us$***

Como la transferencia de Bessel ya estaba normalizada tal que $D(\omega=0) = 1s$ hay que hallar la norma de frecuencia. Para ello hay que tener en cuenta que la normalización se realizó tal que:

$$
\frac{D(\omega=0)}{\Omega_t} = 1s
$$
Por ende:

$$
\Omega_t = 200us
$$

Pero tal $\Omega$ sería temporal y se quiere la de frecuencia, con lo cual:

$$
\Omega_f = \frac{1}{\Omega_t} = 5000 \frac{rad}{s}
$$

### *Sección de 1° orden*

Al ser $\Omega_z = R = 1 k\Omega$ y $\Omega_f = 5000 \frac{rad}{s}$, entonces:

$$
R = R'\cdot \Omega_z = 1k\Omega
$$

$$
C= \frac{C'}{\Omega_z \cdot \Omega_f} = \frac{0.4306}{1k\Omega \cdot 5000 \frac{rad}{s}} = 86.12nF
$$

### *Sección de 2° orden*

Al ser $\Omega_z = R_1=R_2 = 1 k\Omega$ y $\Omega_f = 5000 \frac{rad}{s}$, entonces:

$$
R_1 = R_1'\cdot \Omega_z = 1k\Omega
$$

$$
R_2 = R_2'\cdot \Omega_z = 1k\Omega
$$

$$
C_1= \frac{C_1'}{\Omega_z \cdot \Omega_f} = \frac{0.2847}{1k\Omega \cdot 5000 \frac{rad}{s}} = 56.94nF 
$$

$$
C_2= \frac{C_2'}{\Omega_z \cdot \Omega_f} = \frac{0.5437}{1k\Omega \cdot 5000 \frac{rad}{s}} = 108.74nF 
$$
